# Exploratory Data Analysis

```python
# Overview
print('rows, cols:',df.shape)
print('\ncolumns:')
print(df.columns.tolist())
print('\nNum duplicates:', df.duplicated().sum())
print('\nMissing values (top 20):')
print(df.isna().sum().sort_values(ascending=False).head(20))
```

```python
# Data Cleaning 

# 1. Definitive drops (zero variance + ID)
cols_to_drop = ["EmployeeCount", "Over18", "StandardHours", "EmployeeNumber"]

# 2. Optional: Drop redundant synthetic rate metrics
rate_cols_to_drop = ["DailyRate", "HourlyRate", "MonthlyRate"]

# Apply drop
df_cleaned = df.drop(columns=cols_to_drop + rate_cols_to_drop)
```

```python
# Verify the cleaned shape (leaves 28 highly relevant columns)
print("Cleaned dataset shape:", df_cleaned.shape)
```

```python
# Attrition numeric column
df_cleaned["Attrition_Num"] = df_cleaned["Attrition"].apply(
    lambda x: 1 if x == "Yes" else 0
)

# Age Group Binning
age_bins = [17, 25, 35, 45, 60]
age_labels = ["Under 25", "25-34", "35-44", "45+"]
df_cleaned["AgeGroup"] = pd.cut(
    df_cleaned["Age"], bins=age_bins, labels=age_labels
)

# Distance Category
dist_bins = [0, 5, 15, 50]
dist_labels = ["Close (0-5km)", "Moderate (6-15km)", "Far (15+km)"]
df_cleaned["DistanceCategory"] = pd.cut(
    df_cleaned["DistanceFromHome"], bins=dist_bins, labels=dist_labels
)
```

```python
# Converting Rating Columns to Readable Text

rating_map = {1: "Low", 2: "Medium", 3: "High", 4: "Very High"}

satisfaction_cols = [
    "EnvironmentSatisfaction",
    "JobSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance",
]
for col in satisfaction_cols:
    df_cleaned[col + "_Label"] = df_cleaned[col].map(rating_map)
```

```python
# Check Attrition Rate by Key Metrics

print(
    "Attrition by Department:\n",
    df_cleaned.groupby("Department")["Attrition_Num"].mean() * 100,
)
print(
    "\nAttrition by OverTime:\n",
    df_cleaned.groupby("OverTime")["Attrition_Num"].mean() * 100,
)
print(
    "\nAttrition by JobRole:\n",
    df_cleaned.groupby("JobRole")["Attrition_Num"].mean() * 100,
)
```
